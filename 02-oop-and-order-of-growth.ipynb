```python
class Point:
    def __init__(self,x=0,y=0):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f"[{str(self.x)}, {str(self.y)}]"
```

```python
p1 = Point(10,20) # notice we didn't pass self here
# p1 is a reference variable
```

```python
class Shape:
    def __init__(self,points):
        self.points = points
        
    def __str__(self):
        ret_str = ""
        for i in self.points:
            ret_str += str(i) + "-"
            
        return ret_str
        
```

```python
p1 = Point(10,20)
p2 = Point(30,40)
p3 = Point(50,60)
p = [p1,p2,p3]
sh = Shape(p)
print(sh)
```

```python
class Rectangle:
    def __init__(self,length=0,width=0):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return 2 * self.length + 2 * self.width
    
    def __str__(self):
        return "L: " + str(self.length) + " W: " + str(self.width)
        
```

```python
rect = Rectangle(20,30)
print(rect)
```

```python
class Square(Rectangle):
    def __init__(self,length):
        super().__init__(length,length)
        
    def __str__(self):
        return "Square: " + super().__str__()
```

```python
sq = Square(10)
```

```python
print(sq)
```

```python
def fib(n):
    if n <= 1:
        return 1
    else:
        return fib(n-2)+fib(n-1)
```

```python
def fib2(n):
    a,b = 1,1
    for i in range(n):
        a,b = b,a+b
        
    return a
```

```python
print('fib: ')
for i in range(15):
    print(fib(i),end=", ")
    
print("\n")
print('fib2: ')
for i in range(15):
    print(fib2(i),end=", ")
```

```python
import time

def compute_times(fn,limit):
    l = []
    for i in range(limit):
        start_time = int(round(time.time()*1000))
        _ = fn(i)
        end_time = int(round(time.time()*1000))
        time_taken = end_time-start_time
        l.append(time_taken)
        
    return l
        
```

```python
limit = 30
fib1_times = compute_times(fib,limit)
fib2_times = compute_times(fib2,limit)
print(fib1_times)
print(fib2_times)
```
