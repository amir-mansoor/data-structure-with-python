```python
class Point:
  def __init__(self,x=0,y=0):
    self.x = x
    self.y = y

  def __str__(self):
    return "[" + str(self.x) + "," + str(self.y) + "]"
```

```python
p1 = Point()
print(p1.x)
print("p1", p1)

p2 = Point(2,3)
print(p2.x)
print("p2",p2)
```

```python
class Shape:
  def __init__(self,points):
    self.points = points

  def print_points(self):
    for i in self.points:
      print(i)



p1 = Point(1,2)
p2 = Point(3,4)
p3 = Point(5,6)
points = [p1,p2,p3]

sh = Shape(points)
sh.print_points()
```

```python
class Triangle(Shape):
  pass
```

```python
t = Triangle(points)
```

```python
t.print_points()
```

```python
def get_area(self):
  vertices = self.points
  n = len(vertices)
  a = 0.0
  for i in range(n):
    j = (i + 1) % n
    a += abs(vertices[i].x * vertices[j].y - vertices[j].x * vertices[i].y)
  return a / 2.0

Triangle.get_area = get_area
```

```python
t.get_area()
```

```python
class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * self.length + 2 * self.width

  def __str__(self):
    return "L: " + str(self.length) + " W: " + str(self.width)

rect = Rectangle(2,4)
print(rect)

```

```python
class Square(Rectangle):
  def __init__(self,length):
    super().__init__(length,length)

  def __str__(self):
    return "Square: " + super().__str__()
```

```python
square = Square(4)
print(square)
```

```python
square.area()
```

```python
## Order of Growth
```

```python
def fib1(n):
  if n <= 1:
    return 1

  return fib1(n-1) + fib1(n-2)
```

```python
fib1(3)
```

```python
def fib2(n):
  a = 1
  b = 1

  for i in range(n):
    a,b = b, a+b
  return a
```

```python
fib2(6)
```

```python
print('fib:  ')
for i in range(15):
  print(fib1(i), end=", ")

```

```python
print('fib2:  ')
for i in range(15):
  print(fib2(i), end=", ")

```

```python
import time
def compute_time(fn,limit):
  l = []
  for i in range(limit):
    start_time = int(round(time.time()*1000))
    fn(i)
    end_time = int(round(time.time()*1000))
    time_taken = end_time - start_time
    l.append(time_taken)
  return l
```

```python
limit = 30
fib1_times = compute_time(fib1,limit)
fib2_times = compute_time(fib2,limit)

print(fib1_times)
print(fib2_times)
```
