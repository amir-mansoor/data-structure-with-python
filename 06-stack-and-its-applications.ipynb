```python
class Stack:
    def __init__(self):
        self.l = []
        
    def push(self,val):
        self.l.append(val)
        
    def pop(self):
        val = self.l.pop()
        return val
    
    def peek(self):
        return self.l[-1]
```

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
s.peek()
```

```python
def bracket_matching(string):
    opening = "[{("
    closing = "]})"
    mapping = dict(zip(opening,closing))
    stack = []
    
    for i in string:
        if i not in mapping.values() and i not in mapping.keys():
            continue
            
        if i in mapping:
            stack.append(mapping[i])
            
        elif len(stack) == 0 or i != stack.pop():
            return False
        
    return len(stack) == 0
```

```python
bracket_matching("[]")
```

```python
def dec_to_bin(num):
    stack = []
    while num != 0:
        rem = num % 2
        num = num // 2
        stack.append(rem)
        
    ret_str = ""
    while stack:
        ret_str += str(stack.pop())
        
    return ret_str
```

```python
dec_to_bin(5)
```

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
        
class Stack:
    def __init__(self):
        self.size = 0 
        self.head = None
        
    def push(self,val):
        new_node = Node(val)
        if self.head is None:
            self.head = new_node
            self.size += 1
            return
        
        temp = self.head
        while temp.next is not None:
            temp = temp.next
            
        temp.next = new_node
        self.size += 1
        
    def pop(self):
        if self.head is None:
            raise IndexError("Cannot pop stack is empty")
            
        if self.head.next is None:
            val = self.head.val
            self.head = None
            return val
        
        temp = self.head
        while temp.next is not None:
            prev = temp
            temp = temp.next
            
        prev.next = None
        val = temp.val
        return val
    
    def peek(self):
        if self.size == 0:
            raise Exception("cannot peek stack is empty")
            
        temp = self.head
        while temp.next is not None:
            temp = temp.next
            
        return temp.val
        
            
        
```

```python
my_stack = Stack()
my_stack.push(1)
my_stack.push(2)
my_stack.push(3)
```
