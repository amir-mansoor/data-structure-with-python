```python
class Stack:
    def __init__(self):
        self.l = []
        
    def push(self,val):
        self.l.append(val)
        
    def pop(self):
        val = self.l.pop()
        return val
    
    def peek(self):
        return self.l[-1]
```

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
s.peek()
```

```python
def bracket_matching(string):
    opening = "[{("
    closing = "]})"
    mapping = dict(zip(opening,closing))
    stack = []
    
    for i in string:
        if i not in mapping.values() and i not in mapping.keys():
            continue
            
        if i in mapping:
            stack.append(mapping[i])
            
        elif len(stack) == 0 or i != stack.pop():
            return False
        
    return len(stack) == 0
```

```python
bracket_matching("[]")
```

```python
def dec_to_bin(num):
    stack = []
    while num != 0:
        rem = num % 2
        num = num // 2
        stack.append(rem)
        
    ret_str = ""
    while stack:
        ret_str += str(stack.pop())
        
    return ret_str
```

```python
dec_to_bin(4)
```
