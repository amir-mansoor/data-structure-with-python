```python
def improve_guess(x,guess):
  return (guess+x/guess)/2

def good_enough(guess,x):
  return abs(guess*guess-x) < 0.00000000001

def sqrt(x,guess=0.0001):
  if good_enough(guess,x):
    return guess
  else:
    return sqrt(x,improve_guess(x,guess))
```

```python
sqrt(25)
```
