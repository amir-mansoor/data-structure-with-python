```python
class Queue:
    def __init__(self):
        self.isEmpty = True
        self.isFull = False
        self.i = 0
        self.o = 0
        self.size = 5
        self.q = list(range(self.size))
```

```python
q = Queue()
```

```python
def _inc(self,val):
    if val + 1 == self.size:
        return 0
    else:
        return val+1
    
Queue._inc = _inc
```

```python
def enqueue(self,val):
    if self.isFull:
        raise IndexError("Cannot enqueue. Queue is full")
        
    self.q[self.i] = val
    self.i = self._inc(self.i)
    if self.i == self.o:
        self.isFull = True
        
    self.isEmpty = False

Queue.enqueue = enqueue
```

```python
def dequeue(self):
    if self.isEmpty:
        raise IndexError("Cannot dequeue. Queue is empty")
        
    ret = self.q[self.o]
    self.o = self._inc(self.o)
    if self.i == self.o:
        self.isEmpty = True
        
    self.isFull = False
    return ret
Queue.dequeue = dequeue
```

```python
def __str__(self):
    return str(self.q) + " , in: " + str(self.i) + ", out: " + str(self.o)

Queue.__str__ = __str__
```
