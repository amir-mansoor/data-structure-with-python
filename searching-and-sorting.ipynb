```python
## Linear Search

def linear_search(lst,target):
    for i in lst:
        if i == target:
            return i
        
    return 0

lst = [1,2,3,4,5]
found = linear_search(lst,5)
if found:
    print("Found")
else:
    print("not found")
```

```python
def binary_search(lst,target):
    low = 0
    high = len(lst) - 1
    while low <= high:
        mid = (low+high)//2
        
        if lst[mid] == target:
            return mid
        
        elif lst[mid] < target:
            low = mid+1
        else:
            high = mid-1
            
        
    return 0

lst = [1,2,3,4,5]
# found = binary_search(lst,5)
found = binary_search(["mansoor","sannan","nauman","khizar"],"nauman")
if found:
    print("Found")
else:
    print("not found")
```

```python
# bubble sort
def bubble_sort(lst):
    for i in range(len(lst)-1):
        swap = False
        for j in range(len(lst)-i-1):
            print(f"run {j} times")
            if lst[j] > lst[j+1]:
                temp = lst[j]
                lst[j] = lst[j+1]
                lst[j+1] = temp
                swap = True
                
        if swap is False:
            return
```

```python
lst = [1,2,3,4,5]
bubble_sort(lst)
lst
```

```python
# Selection sort
def selection_sort(lst):
    for i in range(len(lst)-1):
        minimum = i
        for j in range(i+1,(len(lst))):
            if lst[j] < lst[minimum]:
                minimum = j
                
        lst[i],lst[minimum] = lst[minimum],lst[i]
        
        
lst = [1,4,6,2,5,3]
selection_sort(lst)
lst
```
