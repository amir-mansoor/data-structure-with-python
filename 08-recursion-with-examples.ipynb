```python
def fac(n):
    if n <= 1:
        return n
    
    return n * fac(n-1)
```

```python
fac(5)
```

```python
for i in range(1,7):
    print(f"Factorial of {i} is: {fac(i)}")
```

```python
def fib1(n):
    if n <= 1:
        return 1
    
    return fib1(n-1) + fib1(n-2)
```

```python
fib1(25)
```

```python
def fib2(n):
    a = 0
    b = 1

    for i in range(n):
        a, b = b, a + b

    return a

fib2(6)
```

```python
%timeit fib2(20)
```

```python
%timeit fib1(20)
```
