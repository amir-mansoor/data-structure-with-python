```python
def fact(n):
  if n <= 1:
    return 1

  return n * fact(n-1)
```

```python
for i in range(0,7):
  print(f"factorial of {i}: ", fact(i) )
```

```python
def fib(n):
  if n <= 1:
    return 1

  return fib(n-1) + fib(n-2)
```

```python
print(fib(10))
```

```python
%timeit fib(20)
```

```python
def fib(n):
  a,b = 0,1
  for i in range(n):
    a,b = b, a+b

  return a

```

```python
%timeit fib(20)
```

```python
print(fib(20))
```

```python
def fib(n, a = 0, b = 1):
  if n == 0:
    return a

  return fib(n-1,b,a+b)
```

```python
%timeit fib(20)
```

```python
def sum_list(l):
  sum = 0
  for i in l:
    sum += i

  return sum
```

```python
l = [1,2,3,4,5]
print(sum_list(l))
```
