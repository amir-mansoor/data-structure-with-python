```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None
        
class Ring:
    def __init__(self):
        self.head = None
```

```python
def __str__(self):
    
    ret_str = "["
    temp = self.head
    
    while temp is not None:
        ret_str += str(temp.val) + ","
        temp = temp.next
        
        if temp == self.head:
            break
            
    ret_str = ret_str.rstrip(",")
    ret_str += "]"
    return ret_str

Ring.__str__ = __str__
```

```python
l = Ring()
print(l)
```

```python
def _get_last(self):
    if self.head is None:
        return None
    
    if self.head.next == self.head:
        return self.head
    
    temp = self.head.next

    while temp.next != self.head:
        temp = temp.next
        
    return temp

Ring._get_last = _get_last
```

```python
def insert(self,index,val):
    
    if self.head is None and index > 0:
        raise IndexError(f"Cannot insert at {index} because list is empty")
    
    new_node = Node(val)
    
    last = self._get_last()
    if index == 0:
        new_node.next = self.head
        self.head = new_node
        
        if last is None:
            self.head.next = self.head
        else:
            last.next = self.head
                
        return    
        
    temp = self.head
    counter = 0
    while temp is not None and counter < index: # temp will never be none
        prev = temp
        temp = temp.next
        counter += 1
        
    prev.next = new_node
    new_node.next = temp
    
    
Ring.insert = insert
```

```python
l = Ring()
l.insert(0,1)
l.insert(1000,2)
l.insert(1,3)
print(l)
```

```python
l._get_last().next == l.head
```

```python
def remove(self,val):
    if self.head is None:
        return None

    
    last = self._get_last()
    temp = self.head
    
    if temp.val == val:
        if last == self.head:
            self.head = None
        else:
            self.head = temp.next
            last.next = self.head
            
        return
    

    prev = temp
    temp = temp.next
    
    while temp != self.head:
        if temp.val == val:
            break
            
        prev = temp
        temp = temp.next
        
    if temp == last.next:
        return 
    
    prev.next = temp.next
    
Ring.remove = remove
    
    
    
        
    
    
```

```python
l = Ring()
l.insert(0,1)
l.insert(1,2)
l.remove(2)
print(l)
```
