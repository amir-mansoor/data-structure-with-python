```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None
        
class Ring:
    def __init__(self):
        self.head = None
```

```python
def _get_last(self):
    if self.head is None:
        return None
    
    if self.head.next == self.head:
        return self.head
    
    temp = self.head.next
    while temp.next != self.head:
        temp = temp.next
    
    return temp
    

Ring._get_last = _get_last
```

```python
def insert(self,index,val):
    
    if self.head is None and index > 0:
        raise IndexError(f"cannot insert at {index}.list is empty")
        
    new_node = Node(val)
    last = self._get_last()
    if index == 0:
        new_node.next = self.head
        self.head = new_node
        if last is None:
            new_node.next = self.head
        else:
            last.next = new_node
        
        return
    
    
    temp = self.head
    counter = 0
    while temp is not None and counter < index: # temp will never be none
        prev = temp
        temp = temp.next
        counter += 1
        

    prev.next = new_node
    new_node.next = temp
    
Ring.insert = insert
    
    
```

```python
def __str__(self):
    
    
    temp = self.head
    last = self._get_last()
    ret_str = "["

    while temp:
        
        ret_str += str(temp.val) + ","
        
        if temp == last:
            break
        
        temp = temp.next
        
    ret_str = ret_str.rstrip(",")
    ret_str += "]"
    return ret_str

Ring.__str__ = __str__
        
    
```

```python
def remove(self,val):
    if self.head is None:
        raise Exception("Cannot remove list is empty")
        
    last = self._get_last()
        
    if self.head.val == val and self.head == last:
#         print("Executing if there is only one value and is found first")
        self.head = None
        return
    else:
#         print("Otherwise if value is first but there are other nodes")
        self.head = self.head.next
        last.next = self.head
        return
    
    temp = self.head
    last = self._get_last()
    while temp != last:
        
        if temp.val == val:
            break
        prev = temp
        temp = temp.next
        
    prev.next = temp.next
    last.next = self.head
        
        
Ring.remove = remove
```

```python
def length(self):
    last = self._get_last()
    temp = self.head
    counter = 0
    while temp:
        
        counter += 1
        if temp == last:
            break
            
        temp = temp.next
        
    return counter

Ring.length = length
```

```python
def pop(self):
    
    if self.head is None:
        raise Exception("Cannot pop list is empty")
    
    last = self._get_last()
    if self.head == last:
        self.head = None
        return
    
    temp = self.head
    while temp:
        if temp.next == last:
            temp.next = self.head
            return
        
        temp = temp.next
        
Ring.pop = pop
    
```

```python
def get(self,index):
    
    if self.head is None:
        raise Exception("Cannot pop list is empty")
    
    temp = self.head
    counter = 0
    while temp is not None and counter < index:
        temp = temp.next
        counter += 1
        
    return temp.val


Ring.get = get
```
