```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None
        

        
```

```python
class LinkedList:
    def __init__(self):
        self.head = None
```

# Push Operation
1. There are two cases the first one if there are no nodes create node and point head to it
2. If there are at least one node then traverse the linkedlist and and stop on the last node then point last.next to new node

```python
def push(self,val):
    new_node = Node(val)
    
    if self.head is None:
#         print("Exec: case 1")
        self.head = new_node
        return 
    
    last = self.head
    while last.next is not None:
        last = last.next
    
#     print("Exec: case 2")
    last.next = new_node
    
LinkedList.push = push
```

```python
def __str__(self):
    ret_str = "["
    
    temp = self.head
    while temp:
        ret_str += str(temp.val) + ","
        temp = temp.next
        
    ret_str = ret_str.rstrip(", ")
    ret_str += "]"
    return ret_str

LinkedList.__str__ = __str__
```

```python
l = LinkedList()
for i in range(1,11):
    l.push(i)
print(l)
```

```python
def pop(self):
    if self.head is None:
        raise Exception("Cannot pop.no value")
        
    # there is only one node
    if self.head.next is None:
        val = self.head.val
        self.head = None
        return val
    
    # there are more nodes
    
    temp = self.head
    while temp.next is not None:
        prev = temp
        temp = temp.next
        
    prev.next = None
    return temp.val

LinkedList.pop = pop
        
```

```python
l = LinkedList()
for i in range(1,11):
    l.push(i)
print(l)
for i in range(1,6):
    l.pop()
print(l)
```

```python
def insert(self,index,val):
    new_node = Node(val)
    
    if index == 0 or self.head is None:
        new_node.next = self.head
        self.head = new_node
        return 
    
    # for other indices
    counter = 0
    temp = self.head
    prev = temp
    while temp is not None and counter < index:
        prev = temp
        temp = temp.next
        counter += 1
    
    
    prev.next = new_node
    new_node.next = temp

LinkedList.insert = insert
```

```python
l = LinkedList()
for i in range(1,11):
    l.push(i)
print(l)
for i in range(1,11):
    l.pop()
print(l)
l.insert(5,100)
l.insert(10,200)
l.insert(0,100)
print(l)
```

```python
def remove(self,val):
    temp = self.head
    if temp is not None:
        if temp.val == val:
            self.head = temp.next
            temp = None
            return
        
    while temp is not None:
        if temp.val == val:
            break
            
        prev = temp
        temp = temp.next
        
    if temp is None: # not found
        return 
    
    prev.next = temp.next # just lose the reference to delete the node
    
    
    
    
LinkedList.remove = remove
```

```python
l = LinkedList()
for i in range(1,11):
    l.insert(i,i)
l.remove(6)
print(l)
l.remove(100)
print(l)
```

```python
def length(self):
    temp = self.head
    counter = 0
    while temp is not None:
        counter += 1
        temp = temp.next
        
    return counter

LinkedList.length = length
```

```python
l = LinkedList()
for i in range(1,11):
    l.insert(i,i)
print(l.length())
```

```python
def get(self,index):
    
    length = self.length()
    print(length)
    if index >= length:
        raise IndexError("Index is out of range")

    counter = 0
    temp = self.head
    while counter < index:
        counter += 1
        temp = temp.next
    
    return temp.val
    
LinkedList.get = get
```
