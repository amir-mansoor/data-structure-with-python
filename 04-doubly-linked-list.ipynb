```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None
        self.prev = None
        
class Doubly:
    def __init__(self):
        self.head = None
```

```python
def push(self,val):
    new_node = Node(val)
    if self.head is None:
        self.head = new_node
        return
    
    temp = self.head
    while temp.next is not None:
        temp = temp.next
        
    temp.next = new_node
    new_node.prev = temp
    
Doubly.push = push
```

```python
def __str__(self):
    ret_str = "["
    
    temp = self.head
    while temp is not None:
        ret_str += f"{temp.val}" + ", "
        temp = temp.next
        
    ret_str = ret_str.rstrip(", ")
    ret_str += "]"
    return ret_str

Doubly.__str__ = __str__
```

```python
l = Doubly()
print(l)
l.push(1)
l.push(2)
l.push(3)
l.push(4)
print(l)
```

```python
def pop(self):
    
    if self.head is None:
        raise Exception("List is empty")
    
    if self.head.next is None:
        val = self.head.val 
        self.head = None
        return val
    
    temp = self.head
    while temp.next is not None:
        prev = temp
        temp = temp.next
        
    val = temp.val
    prev.next = None
    return val

Doubly.pop = pop
    
```

```python
l = Doubly()
l.push(1)
l.push(2)
l.push(3)
l.push(4)
l.pop()
print(l)
```

```python
def insert(self,index,val):
    new_node = Node(val)
    if index == 0:
        new_node.next = self.head
        if self.head is not None:
            self.head.prev = new_node
        self.head = new_node
        return 
    
    # case 2
    temp = self.head
    counter = 0
    while temp is not None and counter < index:
        prev = temp
        temp = temp.next
        counter += 1
        
    prev.next = new_node
    new_node.prev = prev
    new_node.next = temp

Doubly.insert = insert
```

```python
l = Doubly()
for i in range(0,6):
    l.insert(i,i)
print(l)
l.insert(6,6)
print(l)
```
