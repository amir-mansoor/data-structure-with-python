```python
class Node:
  def __init__(self,val):
    self.val = val
    self.next = None
    self.prev = None
```

```python
class Doubly:
  def __init__(self):
    self.head = None
```

```python
def push(self,val):
  new_node = Node(val)

  # first case if head is None
  if self.head is None:
    self.head = new_node
    return

  # second case
  last = self.head
  while last.next is not None:
    last = last.next

  last.next = new_node
  new_node.prev = last

Doubly.push = push
```

```python
def __str__(self):
  ret_str = "["

  temp = self.head

  while temp is not None:
    ret_str += str(temp.val) + ", "
    temp = temp.next

  ret_str = ret_str.strip(", ")
  ret_str += "]"

  return ret_str

Doubly.__str__ = __str__
```

```python
def pop(self):
  if self.head is None:
    raise Exception("Cannot pop, no values")

  # if there is only one value
  if self.head.next is None:
    val = self.head.val
    self.head = None # automatic garbage collection
    return val

  temp = self.head

  while temp.next is not None:
    prev = temp
    temp = temp.next

  val = temp.val
  prev.next = None
  return val

Doubly.pop = pop





```

```python
l = Doubly()
l.push(1)
l.push(2)
l.push(3)
# print(l)

# l.pop()
print(l)
```

```python
def insert(self,index,val):

  # insertion at index 0
  new_node = Node(val)
  if index == 0:
    new_node.next = self.head
    if self.head is not None:
      self.head.prev = new_node

    self.head = new_node
    return

  # for other indices
  temp = self.head

  counter = 0
  while temp is not None and counter < index:
    prev = temp
    temp = temp.next

    counter += 1

  prev.next = new_node
  new_node.prev = prev

  new_node.next = temp
  if temp is not None:
    temp.prev = new_node


Doubly.insert = insert


```

```python
l = Doubly()
l.push(1)
l.push(2)
l.push(3)
print(l)
l.insert(0,10)
print(l)
l.insert(4,4)
print(l)
l.insert(5,5)
print(l)
l.insert(10,6)
print(l)
l.insert(2,2)
print(l)
l.insert(8,7)
print(l)
```

```python
def remove(self,val):
  if self.head is None:
    raise Exception("No values")

  if self.head.val == val:
    self.head = self.head.next
    if self.head is not None:
      self.head.prev = None
    return

  temp = self.head
  while temp is not None:
    if temp.val == val:
      break
    else:
      prev = temp
      temp = temp.next

  prev.next = temp.next
  if temp.next is not None:
    temp.next.prev = prev

  # prev = prev.next.next
  # prev.next.next = prev
  return




Doubly.remove = remove
```

```python
l = Doubly()
l.push(1)
l.push(2)
l.push(3)
l.push(20)
print(l)
l.remove(3)
print(l)
l.remove(1)
print(l)
l.remove(20)
print(l)
l.remove(2)

```
