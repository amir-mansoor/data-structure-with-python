```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None
        self.prev = None
```

```python
class Doubly:
    def __init__(self):
        self.head = None
```

```python
def push(self,val):
    new_node = Node(val)
    if self.head is None:
        self.head = new_node
        return 
    
    temp = self.head
    while temp.next is not None:
        temp = temp.next
        
    temp.next = new_node
    new_node.prev = temp
        
Doubly.push = push
```

```python
l = Doubly()
for i in range(0,11):
    l.push(i)
print(l)
```

```python
def __str__(self):
    ret_str = "["
    
    temp = self.head
    while temp:
        ret_str += str(temp.val) + ", "
        temp = temp.next
        
    ret_str = ret_str.strip(", ")
    ret_str += "]"
    return ret_str

Doubly.__str__ = __str__
```

```python
l = Doubly()
for i in range(1,11):
    l.push(i)
    
print(l)
```

```python
def pop(self):
    if self.head is None:
        raise Exception("Cannot pop. no values")
    
    if self.head.next is None:
        self.head = None
        return 
    
    temp = self.head
    while temp.next is not None:
        prev = temp
        temp = temp.next
        
    
    prev.next = None
    
    
Doubly.pop = pop
```

```python
l = Doubly()
for i in range(1,11):
    l.push(i)

l.pop()
print(l)
```

```python
def insert(self,index,val):
    
    if index > self.length():
        raise IndexError("Index is out of range")

    
    new_node = Node(val)
    if index == 0:
        new_node.next = self.head
        
        if self.head is not None:
            self.head.prev = new_node
            
        self.head = new_node
        
        return 
    
    temp = self.head
    counter = 0
    while temp is not None and counter < index:
        prev = temp
        temp = temp.next
        counter += 1
        
    prev.next = new_node
    new_node.prev = prev
        
    if temp is not None:
        new_node.next = temp
        temp.prev = new_node
        
        
Doubly.insert = insert
        
        
    
        
```

```python
l = Doubly()
l.insert(0,1)
l.insert(1,2)
l.insert(2,3)
l.insert(3,4)
l.insert(1,10)
print(l)
```

```python
def length(self):
    counter = 0
    temp = self.head
    while temp:
        counter += 1
        temp = temp.next
        
    return counter

Doubly.length = length
```

```python
l = Doubly()
l.insert(0,1)
l.insert(1,2)
l.insert(2,3)
l.insert(3,4)
l.insert(1,10)
print(l.length())
```

```python
def remove(self,val):
    if self.head.val == val:
        self.head = self.head.next
        return
    
    temp = self.head
    while temp is not None:
        if temp.val == val:
            break
            
        prev = temp
        temp = temp.next
        
    if temp is not None:
        prev.next = temp.next
        if temp.next is not None:
            temp.next.prev = prev
    return

Doubly.remove = remove
```

```python
l = Doubly()
for i in range(0,11):
    l.insert(i,i)
print(l)
l.remove(1)
l.remove(10)
print(l)
```
