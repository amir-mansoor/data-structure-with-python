# Writing Your Own Stack

```python
class Stack:
  def __init__(self):
    self.l = []

  def push(self,val):
    self.l.append(val)

  def pop(self):
    return self.l.pop()

  def peek(self):
    return self.l[-1]

```

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)

print(s)
s.pop()
```

```python
print(s[1]) # <----- error ---- which for us is success -----------> this is an error we want
```

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
print(s.peek())
print(s.pop())
print(s.peek())
print(s.pop())
```

```python
def is_matched(string):
  opening = '{[('
  closing = '}])'
  maping = dict(zip(opening,closing))

  stack = []

  for c in string:
    # case 1
    if c not in maping.values() and c not in maping.keys():
      continue

    # case 2
    if c in maping:
      stack.append(maping[c])

    # case 3
    elif len(stack) == 0 or c != stack.pop():
      return False

  return len(stack) == 0
```

```python
is_matched('[]')
```

```python
is_matched("{()[(]}")
```

```python
is_matched('2 + (3 * 4) + [5 + 3] + 2')
```

```python
def binary(n):
  bits = []
  while n != 0:
    bits.append(n % 2)
    n = n // 2

  while len(bits) != 0:
    print(bits.pop(), end="")

```

```python
binary(9)
```
